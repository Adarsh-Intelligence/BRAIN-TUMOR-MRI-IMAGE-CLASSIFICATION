```

 (gradcam.py)     (evaluate.py)            (train.py: Custom CNN | EfficientNetB0)
```

| Step | Module | Command |
|---|---|---|
| Explore data | `src/explore.py` | `python -m src.explore` |
| Build splits | `src/prepare_data.py` | `python -m src.prepare_data` |
| Train | `src/train.py` | `python -m src.train --model efficientnet` |
| Evaluate | `src/evaluate.py` | `python -m src.evaluate --model efficientnet` |

## 2. Integrity checks & duplicate detection

Before splitting we must answer: *are any images corrupted?* and *are some images
(near-)copies of each other?* `inspect_image()` computes for every file:

* **MD5** of the bytes → identical files
* **pHash** (perceptual hash, 64 bits) → visually near-identical images, even after re-encoding/resizing

Let's look at one image's hashes:

````python
info = D.inspect_image(PROJECT_ROOT / raw.filepath.iloc[0])
print({k: info[k] for k in ("readable", "width", "height", "mode", "md5")})
print("pHash bits:", format(int(info["phash"]), "064b"))
````

### How pHash works (in `data_loader.phash`)

1. grayscale + resize to 32×32
2. 2-D DCT (frequency transform), keep the 8×8 low-frequency block
3. each bit = "is this coefficient above the median?"

Two images are near-duplicates if their 64-bit hashes differ in **≤ 4 bits** (Hamming distance).
Below are real pairs found in the dataset — note they often sit in **different original splits**.

````python
near = pd.read_csv(C.EXPLORATION_DIR / "near_duplicate_pairs.csv")
print(f"near-duplicate pairs: {len(near)}  (grouped: {near.grouped.sum()})")
cross = near[near.split_a != near.split_b].sort_values("hamming").head(4)

fig, axes = plt.subplots(len(cross), 2, figsize=(6, 3 * len(cross)))
for i, (_, r) in enumerate(cross.iterrows()):
    for j, side in enumerate("ab"):
        axes[i, j].imshow(P.load_image(PROJECT_ROOT / r[f"filepath_{side}"]))
        axes[i, j].set_title(f"{r[f'class_{side}']} | orig. {r[f'split_{side}']}", fontsize=9)
        axes[i, j].axis("off")
    # recompute distance live to show it matches
    ha = D.inspect_image(PROJECT_ROOT / r.filepath_a)["phash"]; hb = D.inspect_image(PROJECT_ROOT / r.filepath_b)["phash"]
    axes[i, 0].set_ylabel(f"d={bin(int(ha) ^ int(hb)).count('1')}")
plt.suptitle("Near-duplicates across the ORIGINAL splits = leakage", y=1.0); plt.tight_layout(); plt.show()
````

### Grouping with Union-Find

`assign_groups()` puts images in the same **group** if they share any of:
a source stem, an MD5, or a pHash distance ≤ 4 (≤ 2 if labels differ, to avoid chaining
unrelated scans). A tiny example:

````python
toy = pd.DataFrame({
    "filepath": list("abcde"), "class_name": ["glioma"] * 5, "original_split": ["train"] * 5,
    "source_stem": ["s1", "s1", "s2", "s3", "s4"],      # a,b share the original image
    "md5":         ["m1", "m2", "m3", "m3", "m4"],      # c,d are byte-identical
    "phash":       [0, 2**63, 2**62, 2**61, 2**64 - 1],
})
grouped, _ = D.assign_groups(toy, phash_threshold=0)
grouped[["filepath", "source_stem", "md5", "group_id", "group_size"]]
````

## 3. Leakage-safe split (70 / 15 / 15)

`grouped_stratified_split()` pools **all** images, then uses `StratifiedGroupKFold` with
20 folds → 14 train / 3 val / 3 test folds. Guarantees:

* a whole **group** always lands in exactly one split (no leakage)
* each split keeps roughly the same **class proportions** (stratified)
* deterministic with `seed = 42`

The result is saved as CSV manifests in `artifacts/splits/`.

````python
splits = {s: D.load_split(s) for s in ("train", "val", "test")}
df = pd.concat(splits.values(), ignore_index=True)
D.verify_no_leakage(df)       # raises AssertionError if any group/md5/stem spans 2 splits
print("✅ no leakage across splits\n")

tab = pd.crosstab(df.class_name, df.split)[["train", "val", "test"]]
display(tab.assign(total=tab.sum(1)))
display((tab / tab.sum()).round(3).style.set_caption("class proportions within each split"))
````

## 6. The two models

### 6a. Custom CNN (from scratch) — `src/models/custom_cnn.py`

VGG-style: 4 blocks of `[Conv → BatchNorm → ReLU] × 2 → MaxPool → Dropout`
(32 → 64 → 128 → 256 filters), then GlobalAveragePooling → Dense(128) → Dropout → Softmax(4).

````python
from src.models.custom_cnn import build_custom_cnn

cnn = build_custom_cnn(num_classes=len(class_names))
cnn.summary(line_length=100)
````

### 6b. EfficientNetB0 transfer learning — `src/models/efficientnet.py`

ImageNet-pretrained backbone + GAP + Dropout + Softmax head, trained in **two stages**:

| Stage | What is trainable | LR | Epochs |
|---|---|---|---|
| A — feature extraction | only the new head (backbone frozen) | 1e-3 | ≤ 15 |
| B — fine-tuning | blocks 6–7 + head (**BatchNorm stays frozen**) | 1e-5 | ≤ 20 |

The stage with the better **validation macro-F1** is kept (fine-tuning is not assumed to help).
Below we build it with `weights=None` just to inspect the architecture (no download).

````python
from src.models.efficientnet import build_efficientnet, unfreeze_top

def n_trainable(m):
    return sum(int(np.prod(w.shape)) for w in m.trainable_weights)

eff = build_efficientnet(len(class_names), weights=None)
print(f"total params           : {eff.count_params():>10,}")
print(f"stage A trainable      : {n_trainable(eff):>10,}  (head only)")
n_layers = unfreeze_top(eff, cfg.efficientnet.finetune_unfreeze_from)
print(f"stage B trainable      : {n_trainable(eff):>10,}  ({n_layers} backbone layers unfrozen)")
print(f"custom CNN params      : {cnn.count_params():>10,}")
````

## 7. Training (`src/train.py`)

Training is done from the command line (it can take a long time — a GPU / WSL2 is recommended):

```bash
python -m src.train --model custom_cnn           # ~40 epochs max
python -m src.train --model efficientnet         # stage A + stage B
python -m src.train --model custom_cnn --smoke   # tiny 1-epoch sanity run on CPU
```

Key ingredients:

* **Loss**: sparse categorical cross-entropy + class weights · **Optimizer**: Adam
* **Callbacks**: `ModelCheckpoint` (best `val_loss`), `EarlyStopping` (restore best weights),
  `ReduceLROnPlateau` (×0.3 after 3 flat epochs), `CSVLogger`
* Model selection uses **validation only** — the test set is never touched here.

Outputs go to `artifacts/models/<model>/`: `model.keras`, `class_mapping.json`,
`preprocessing.json`, `training_summary.json`, history CSV/JSON, and plots.

````python
trained = sorted(p.parent.name for p in C.MODELS_DIR.glob("*/model.keras"))
print("Trained models found:", trained or "none yet")
for name in trained:
    s = json.loads((C.MODELS_DIR / name / "training_summary.json").read_text())
    print(f"\n{name}: smoke={s['smoke']} train={s['train_size']} val={s['val_size']} "
          f"selected={s['selected_stage']} time={s['train_time_s']:.0f}s params={s['params']:,}")
    display(pd.read_csv(sorted((C.MODELS_DIR / name).glob("*_log.csv"))[0]).round(4))
````

Reported on the **held-out test split**: accuracy, macro / weighted precision-recall-F1,

**Why macro-F1?** With class imbalance, accuracy can hide a weak class. Macro-F1 averages
F1 over classes equally, so every tumour type counts the same. Toy example:

````python
from src.evaluate import compute_metrics

y_true = [0, 0, 0, 0, 1, 1, 2, 2, 3, 3]
y_pred = [0, 0, 0, 0, 0, 1, 2, 3, 3, 3]
m = compute_metrics(y_true, y_pred, class_names)
print(m["classification_report"])
print("confusion matrix (rows = true, cols = predicted):")
pd.DataFrame(m["confusion_matrix"], index=class_names, columns=class_names)
````

## 9. Grad-CAM — what is the model looking at? (`src/gradcam.py`)

Grad-CAM weights the last convolutional feature map by the gradient of the predicted
class score → a coarse heatmap of the regions that **increased** that score.

⚠️ It is an *attribution*, **not** a tumour segmentation — it can also reveal shortcuts
(e.g. focusing on the skull or text overlays).

Below we run it on whatever trained model is available (prefers the real models over the smoke one).

````python
from src.gradcam import gradcam_heatmap, overlay
%matplotlib inline

pref = [n for n in ("efficientnet", "custom_cnn", "smoke_custom_cnn") if n in trained]
if not pref:
    print("No trained model yet — train one first.")
else:
    model_name = pref[0]
    model = keras.models.load_model(C.MODELS_DIR / model_name / "model.keras")
    sample = splits["test"].groupby("class_name").head(1)
    fig, axes = plt.subplots(2, len(sample), figsize=(3 * len(sample), 6.3))
    for i, (_, r) in enumerate(sample.iterrows()):
        img = P.load_image(PROJECT_ROOT / r.filepath)
        cam, ci, probs = gradcam_heatmap(model, img.astype("float32"))
        ok = class_names[ci] == r.class_name
        axes[0, i].imshow(img); axes[0, i].set_title(f"true: {r.class_name}", fontsize=9)
        axes[1, i].imshow(overlay(img, cam))
        axes[1, i].set_title(f"pred: {class_names[ci]} ({probs[ci]:.2f})", fontsize=9, color="green" if ok else "red")
        for a in axes[:, i]:
            a.axis("off")
    plt.suptitle(f"Grad-CAM — {model_name}", y=1.0); plt.tight_layout(); plt.show()
````

## 10. Predicting a single image (how an app would use the model)

````python
def predict(path, model_name=None):
    model_name = model_name or pref[0]
    mdl = keras.models.load_model(C.MODELS_DIR / model_name / "model.keras")
    names = json.loads((C.MODELS_DIR / model_name / "class_mapping.json").read_text())["class_names"]
    x = P.prepare_batch([path])                      # (1, 224, 224, 3) float32 in [0, 255]
    probs = mdl.predict(x, verbose=0)[0]
    return pd.Series(probs, index=names).sort_values(ascending=False)

if pref:
    example = PROJECT_ROOT / splits["test"].filepath.iloc[0]
    print("true label:", splits["test"].class_name.iloc[0])
    display(predict(example).round(4).to_frame("probability"))
````

## ✅ Recap

2. **Inspect** every file (corruption, MD5, pHash) — 0 corrupted, 0 exact copies, but many near-duplicates.

6. **Augment** mildly, training data only; use class weights for imbalance.
7. **Train** a Custom CNN and EfficientNetB0 (2-stage), selecting on **validation** macro-F1.
8. **Evaluate** once on **test**; compare accuracy, macro-F1, size and speed.
9. **Explain** predictions with Grad-CAM (attribution, not segmentation).
